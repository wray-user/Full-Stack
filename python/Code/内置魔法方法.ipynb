# 1. 什么是魔法方法
魔法方法是Python中以双下划线（__）开头和结尾的特殊方法。它们是Python类的一部分，用于定义类的行为和功能。魔法方法不需要直接调用，而是由Python解释器在特定场景下调用。通过实现这些方法，可以让你的类更像内置类型，从而提高代码的可读性和可维护性。

# 2. 初始化与销毁
    __init__
    __del__

     __init__ 是类的构造方法，用于初始化对象。当你创建一个类的实例时，Python会自动调用 __init__ 方法。
     __del__ 是类的析构方法，用于在对象被销毁时执行清理操作。当对象的引用计数为0时，Python会自动调用 __del__ 方法。


```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

# 创建对象
person = Person("Alice", 25)
print(person.name)  # 输出: Alice
print(person.age)   # 输出: 25
```

```python
class Person:
    def __init__(self, name):
        self.name = name

    def __del__(self):
        print(f"Object {self.name} is being destroyed")

# 创建对象
person = Person("Alice")
del person  # 输出: Object Alice is being destroyed
```

# 3. 运算符重载
运算符重载允许你为自定义类定义运算符的行为。通过实现特定的魔法方法，可以让类支持各种运算符操作。

     __add__, __sub__, __mul__, __truediv__
    这些方法分别用于重载加法、减法、乘法和除法运算符


    __iadd__, __isub__, __imul__, __itruediv__
    这些方法用于实现原地运算符（如 +=, -=, *=, /=）。

    __eq__, __ne__, __lt__, __gt__, __le__, __ge__
    这些方法用于重载比较运算符（如 ==, !=, <, >, <=, >=）。

    __neg__, __pos__, __abs__
    这些方法用于重载负号、正号和绝对值运算符。

    __len__, __getitem__, __setitem__, __delitem__
    这些方法用于实现容器协议，使类支持长度计算和索引操作。

```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    def __sub__(self, other):
        return Vector(self.x - other.x, self.y - other.y)

    def __mul__(self, scalar):
        return Vector(self.x * scalar, self.y * scalar)

    def __truediv__(self, scalar):
        return Vector(self.x / scalar, self.y / scalar)

# 创建对象
v1 = Vector(1, 2)
v2 = Vector(3, 4)

# 加法
v3 = v1 + v2
print(v3.x, v3.y)  # 输出: 4 6

# 减法
v4 = v1 - v2
print(v4.x, v4.y)  # 输出: -2 -2

# 乘法
v5 = v1 * 2
print(v5.x, v5.y)  # 输出: 2 4

# 除法
v6 = v1 / 2
print(v6.x, v6.y)  # 输出: 0.5 1.0
```

```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __iadd__(self, other):
        self.x += other.x
        self.y += other.y
        return self

    def __isub__(self, other):
        self.x -= other.x
        self.y -= other.y
        return self

    def __imul__(self, scalar):
        self.x *= scalar
        self.y *= scalar
        return self

    def __itruediv__(self, scalar):
        self.x /= scalar
        self.y /= scalar
        return self

# 创建对象
v1 = Vector(1, 2)
v2 = Vector(3, 4)

# 原地加法
v1 += v2
print(v1.x, v1.y)  # 输出: 4 6

# 原地减法
v1 -= v2
print(v1.x, v1.y)  # 输出: 1 2

# 原地乘法
v1 *= 2
print(v1.x, v1.y)  # 输出: 2 4

# 原地除法
v1 /= 2
print(v1.x, v1.y)  # 输出: 1.0 2.0
```

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __eq__(self, other):
        return self.age == other.age

    def __lt__(self, other):
        return self.age < other.age

    def __gt__(self, other):
        return self.age > other.age

# 创建对象
p1 = Person("Alice", 25)
p2 = Person("Bob", 30)

print(p1 == p2)  # 输出: False
print(p1 < p2)   # 输出: True
print(p1 > p2)   # 输出: False
```

```python
class Number:
    def __init__(self, value):
        self.value = value

    def __neg__(self):
        return Number(-self.value)

    def __pos__(self):
        return Number(self.value)

    def __abs__(self):
        return Number(abs(self.value))

# 创建对象
num = Number(-5)

print(-num.value)  # 输出: 5
print(+num.value)  # 输出: -5
print(abs(num.value))  # 输出: 5
```

```python
class MyList:
    def __init__(self):
        self.items = []

    def __len__(self):
        return len(self.items)

    def __getitem__(self, index):
        return self.items[index]

    def __setitem__(self, index, value):
        self.items[index] = value

    def __delitem__(self, index):
        del self.items[index]

# 创建对象
my_list = MyList()
my_list.items = [1, 2, 3, 4, 5]

print(len(my_list))  # 输出: 5
print(my_list[0])    # 输出: 1
my_list[1] = 10
print(my_list[1])    # 输出: 10
del my_list[2]
print(my_list.items) # 输出: [1, 10, 4, 5]
```

# 4. 容器协议
容器协议使类能够表现得像内置容器类型（如列表、字典等）。通过实现特定的魔法方法，可以让类支持迭代、索引和成员检查等操作。

    __iter__, __reversed__, __contains__
    这些方法用于实现迭代器协议和成员检查。

```python
class MyList:
    def __init__(self):
        self.items = [1, 2, 3, 4, 5]

    def __iter__(self):
        return iter(self.items)

    def __reversed__(self):
        return reversed(self.items)

    def __contains__(self, item):
        return item in self.items

# 创建对象
my_list = MyList()

# 迭代
for item in my_list:
    print(item)  # 输出: 1, 2, 3, 4, 5

# 反向迭代
for item in reversed(my_list):
    print(item)  # 输出: 5, 4, 3, 2, 1

# 成员检查
print(3 in my_list)  # 输出: True
print(6 in my_list)  # 输出: False
```

# 5. 上下文管理器
上下文管理器允许你定义资源的获取和释放行为，通常用于文件操作、数据库连接等场景。

    __enter__, __exit__
    这些方法用于实现上下文管理器。

```python
class FileHandler:
    def __init__(self, filename, mode):
        self.filename = filename
        self.mode = mode
        self.file = None

    def __enter__(self):
        self.file = open(self.filename, self.mode)
        return self.file

    def __exit__(self, exc_type, exc_val, exc_tb):
        if self.file:
            self.file.close()

# 使用上下文管理器
with FileHandler("example.txt", "w") as f:
    f.write("Hello, World!")
```

# 6. 字符串表示
字符串表示方法用于定义对象的字符串表示形式，方便调试和输出。
    
    __str__, __repr__, __format__
    这些方法用于定义对象的字符串表示形式。

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"Person(name={self.name}, age={self.age})"

    def __repr__(self):
        return f"Person('{self.name}', {self.age})"

    def __format__(self, format_spec):
        if format_spec == "short":
            return f"{self.name} ({self.age})"
        return str(self)

# 创建对象
person = Person("Alice", 25)

print(str(person))       # 输出: Person(name=Alice, age=25)
print(repr(person))      # 输出: Person('Alice', 25)
print(f"{person:short}") # 输出: Alice (25)
```

# 7. 类型转换
类型转换方法用于定义对象的类型转换行为。

    __int__, __float__, __complex__, __bool__
    这些方法用于定义对象的数值类型转换。

    __bytes__, __hash__, __call__
    这些方法用于定义对象的字节表示、哈希值和可调用行为。

```python
class Number:
    def __init__(self, value):
        self.value = value

    def __int__(self):
        return int(self.value)

    def __float__(self):
        return float(self.value)

    def __complex__(self):
        return complex(self.value)

    def __bool__(self):
        return bool(self.value)

# 创建对象
num = Number(5)

print(int(num))   # 输出: 5
print(float(num)) # 输出: 5.0
print(complex(num)) # 输出: (5+0j)
print(bool(num))  # 输出: True
```

```python
class MyObject:
    def __init__(self, value):
        self.value = value

    def __bytes__(self):
        return bytes(self.value, "utf-8")

    def __hash__(self):
        return hash(self.value)

    def __call__(self):
        return f"Called with value: {self.value}"

# 创建对象
obj = MyObject("Hello")

print(bytes(obj))  # 输出: b'Hello'
print(hash(obj))   # 输出: 哈希值
print(obj())       # 输出: Called with value: Hello
```

#  8. 属性访问
属性访问方法用于定义对象的属性访问行为。

     __getattr__, __getattribute__, __setattr__, __delattr__
    这些方法用于定义对象的属性访问、设置和删除行为。

```python
class Person:
    def __init__(self, name):
        self.name = name

    def __getattr__(self, attr):
        return f"Attribute {attr} not found"

    def __getattribute__(self, attr):
        return super().__getattribute__(attr)

    def __setattr__(self, attr, value):
        self.__dict__[attr] = value

    def __delattr__(self, attr):
        del self.__dict__[attr]

# 创建对象
person = Person("Alice")

print(person.name)    # 输出: Alice
print(person.age)     # 输出: Attribute age not found
person.age = 25
print(person.age)     # 输出: 25
del person.age
print(person.age)     # 输出: Attribute age not found
```

# 9. 数学运算
这些方法用于定义对象的加法、减法、乘法、除法、取模和幂运算。

    __add__, __sub__, __mul__, __truediv__, __floordiv__, __mod__, __pow__
    这些方法用于定义对象的加法、减法、乘法、除法、取模和幂运算。

    __lshift__, __rshift__, __and__, __or__, __xor__
    这些方法用于定义对象的位运算。

```python
class Number:
    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return Number(self.value + other.value)

    def __sub__(self, other):
        return Number(self.value - other.value)

    def __mul__(self, other):
        return Number(self.value * other.value)

    def __truediv__(self, other):
        return Number(self.value / other.value)

    def __floordiv__(self, other):
        return Number(self.value // other.value)

    def __mod__(self, other):
        return Number(self.value % other.value)

    def __pow__(self, other):
        return Number(self.value ** other.value)

# 创建对象
num1 = Number(5)
num2 = Number(2)

print((num1 + num2).value)  # 输出: 7
print((num1 - num2).value)  # 输出: 3
print((num1 * num2).value)  # 输出: 10
print((num1 / num2).value)  # 输出: 2.5
print((num1 // num2).value) # 输出: 2
print((num1 % num2).value)  # 输出: 1
print((num1 ** num2).value) # 输出: 25
```

```python
class Number:
    def __init__(self, value):
        self.value = value

    def __lshift__(self, other):
        return Number(self.value << other.value)

    def __rshift__(self, other):
        return Number(self.value >> other.value)

    def __and__(self, other):
        return Number(self.value & other.value)

    def __or__(self, other):
        return Number(self.value | other.value)

    def __xor__(self, other):
        return Number(self.value ^ other.value)

# 创建对象
num1 = Number(5)  # 二进制: 101
num2 = Number(3)  # 二进制: 011

print((num1 << num2).value) # 输出: 20 (101 << 011 = 10100)
print((num1 >> num2).value) # 输出: 1  (101 >> 011 = 001)
print((num1 & num2).value)  # 输出: 1  (101 & 011 = 001)
print((num1 | num2).value)  # 输出: 7  (101 | 011 = 111)
print((num1 ^ num2).value)  # 输出: 6  (101 ^ 011 = 110)
```

# 10. 其他魔法方法
除了上述常见的魔法方法，还有一些其他的方法用于定义对象的特定行为。

    __dir__, __doc__, __name__, __module__, __class__
    这些方法用于定义对象的目录、文档字符串、名称、模块和类信息。

```python
class MyClass:
    def __init__(self):
        self.value = 42

    def __dir__(self):
        return super().__dir__() + ["custom_attribute"]

    def __doc__(self):
        return "This is a custom class"

    def __name__(self):
        return "MyClass"

    def __module__(self):
        return "__main__"

    def __class__(self):
        return MyClass

# 创建对象
obj = MyClass()

print(dir(obj))       # 输出: ['custom_attribute', ...]
print(obj.__doc__)    # 输出: This is a custom class
print(obj.__name__)   # 输出: MyClass
print(obj.__module__) # 输出: __main__
print(obj.__class__)  # 输出: <class '__main__.MyClass'>
```
